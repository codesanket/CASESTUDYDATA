```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('employees_data.csv')

print(df.head(5))
```

#Load the dataset

```python
df=pd.read_csv('employees_data.csv')
#shape of dataset

print('the shape of Dataset',df.shape)

```

#information about DataSet


```python
print('display information about Dataset :')

print(df.info())
```

#Descritive stat

```python
print('the descriptive statistics of Employee data:')
print(df.describe())
#check missing values and count
```

#check missing values and count

```python
print('the missing values of Employee data:')
print(df.isnull().sum())
```

#Remove duplicate from Dataset

```python
print('check duplicate')
print(df.duplicated())
```

#employee count by job title


```python
jtitle=df['JobTitle'].value_counts()
print(jtitle)
```

```python
jtitle.plot(kind='bar')
plt.title('job count by title')
plt.xlabel('job title')
plt.ylabel('job count')
plt.grid(color='green',linestyle='--',linewidth=1)
plt.tight_layout()
plt.show()
```

#Avg salary by job title


```python
avsalary=df.groupby('JobTitle')['Salary'].mean()
print(avsalary)
```

```python
avsalary.plot(kind='bar')
plt.title('Avg salary by title')
plt.xlabel('job title')
plt.ylabel('Avg salary')
plt.grid(color='green',linestyle='--',linewidth=1)
plt.tight_layout()
plt.show()
```

#Gender distribuction

```python
gender=df['Gender'].value_counts()
print(gender)
```

```python
gender.plot(kind='bar')
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.grid(color='green',linestyle='--',linewidth=1)
plt.tight_layout()
plt.show()
```

#Age Distribuction

```python
age=df['Age'].value_counts()
print(age)
```

```python
age.hist(bins=20)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('employee')
plt.tight_layout()
plt.show()
```

#Average salary by gender

```python
gensalary=df.groupby('Gender')['Salary'].mean()
print(gensalary)
```

```python
gensalary.plot(kind='bar')
plt.title('Avg salary by gender')
plt.xlabel('Gender')
plt.ylabel('Avg salary')
plt.grid(color='green',linestyle='--',linewidth=1)
plt.tight_layout()
plt.show()
```

#avg salary by location


```python
locsalary=df.groupby('Location')['Salary'].mean()
print(locsalary)
```

```python
locsalary.plot(kind='bar')
plt.title('Salary By Location')
plt.xlabel('location')
plt.ylabel('highsalary')
plt.grid(color='green',linestyle='--',linewidth=1)
plt.tight_layout()
plt.show()

```

#Employee count by location

```python
eloc=df['Location'].value_counts()
print(eloc)
```

```python
eloc.plot(kind='bar')
plt.title('Employee count by location')
plt.xlabel('location')
plt.ylabel('count')
plt.grid(color='green',linestyle='--',linewidth=0.5)
plt.tight_layout()
plt.show()
```
